```python
import mysql.connector

# ==========================================================
# KONEKSI — pakai database logika_fuzzy
# ==========================================================
db = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",  # isi password MySQL kamu
    database="logika_fuzzy"
)
print("Koneksi ke database logika_fuzzy berhasil!\n")

# ==========================================================
# FUNGSI — NAMA HARUS SAMA PERSIS DENGAN DI SQL
# ==========================================================
# --- Bayi ---
def naik_bayi(x):
    return x / 2
def trapesium_down_bayi(x):
    return (5 - x) / 2

# --- Anak ---
def naik_anak(x):
    return (x - 6) / 2
def trapesium_down_anak(x):
    return (11 - x) / 2

# --- Remaja ---
def naik_remaja(x):
    return (x - 10) / 4
def trapesium_down_remaja(x):
    return (19 - x) / 4

# --- Pemuda ---
def naik_pemuda(x):
    return (x - 15) / 4
def trapesium_down_pemuda(x):
    return (24 - x) / 4

# --- Dewasa ---
def naik_dewasa(x):
    return (x - 20) / 22
def trapesium_down_dewasa(x):
    return (65 - x) / 22

# --- Lansia ---
def naik_lansia(x):
    return (x - 60) / 9.5
def trapesium_down_lansia(x):
    return (80 - x) / 9.5

# Peta — SAMA PERSIS dengan kolom fungsi di SQL
peta = {
    "naik_bayi": naik_bayi,
    "trapesium_down_bayi": trapesium_down_bayi,
    "naik_anak": naik_anak,
    "trapesium_down_anak": trapesium_down_anak,
    "naik_remaja": naik_remaja,
    "trapesium_down_remaja": trapesium_down_remaja,
    "naik_pemuda": naik_pemuda,
    "trapesium_down_pemuda": trapesium_down_pemuda,
    "naik_dewasa": naik_dewasa,
    "trapesium_down_dewasa": trapesium_down_dewasa,
    "naik_lansia": naik_lansia,
    "trapesium_down_lansia": trapesium_down_lansia
}

# ==========================================================
# FUNGSI AMBIL NILAI
# ==========================================================
def hitung(x, nama_tabel):
    cursor = db.cursor()
    query = f"""
        SELECT b_bawah, b_atas, fungsi
        FROM {nama_tabel}
        WHERE %s >= b_bawah AND %s < b_atas
        LIMIT 1
    """
    cursor.execute(query, (x, x))
    hasil = cursor.fetchone()
    cursor.close()

    if not hasil:
        return 0.0

    _, _, nama_fungsi = hasil

    if nama_fungsi == "0":
        return 0.0
    if nama_fungsi == "1":
        return 1.0
    if nama_fungsi in peta:
        return round(peta[nama_fungsi](x), 4)
    
    raise ValueError(f"Fungsi '{nama_fungsi}' tidak ditemukan!")

# ==========================================================
# FUZZIFIKASI LENGKAP
# ==========================================================
def fuzzifikasi(usia):
    return {
        "Bayi": hitung(usia, "bayi"),
        "Anak": hitung(usia, "anak"),
        "Remaja": hitung(usia, "remaja"),
        "Pemuda": hitung(usia, "pemuda"),
        "Dewasa": hitung(usia, "dewasa"),
        "Lansia": hitung(usia, "lansia")
    }

# ==========================================================
# PENGUJIAN
# ==========================================================
daftar_usia = [1, 4, 7, 12, 17, 22, 30, 50, 65, 75]

print("="*85)
print(f"{'No':<3} {'Usia':<6} {'Bayi':<8} {'Anak':<8} {'Remaja':<8} {'Pemuda':<8} {'Dewasa':<8} {'Lansia':<8}")
print("-"*85)

for no, u in enumerate(daftar_usia, 1):
    h = fuzzifikasi(u)
    print(f"{no:<3} {u:<6} {h['Bayi']:<8.4f} {h['Anak']:<8.4f} {h['Remaja']:<8.4f} "
          f"{h['Pemuda']:<8.4f} {h['Dewasa']:<8.4f} {h['Lansia']:<8.4f}")

print("="*85)
db.close()
```

```python
import mysql.connector

# 1. KONEKSI
db = mysql.connector.connect(
    host="localhost",
    user="root",
    password="",       # WampServer: biarkan kosong
    database="logika_fuzzy_praktikum3"
)
print("Koneksi ke database logika_fuzzy_praktikum3 berhasil!\n")

# 2. FUNGSI — SAMA PERSIS DENGAN TUGAS 1
def naik_bayi(x):     return x / 2.5
def turun_bayi(x):    return (5 - x) / 2.5

def naik_anak(x):     return (x - 6) / 2.5
def turun_anak(x):    return (11 - x) / 2.5

def naik_remaja(x):   return (x - 10) / 4.5
def turun_remaja(x):  return (19 - x) / 4.5

def naik_pemuda(x):   return (x - 15) / 4.5
def turun_pemuda(x):  return (24 - x) / 4.5

def naik_dewasa(x):   return (x - 20) / 22.5
def turun_dewasa(x):  return (65 - x) / 22.5

def naik_lansia(x):   return (x - 60) / 10
def turun_lansia(x):  return (80 - x) / 10

# 3. PETA — NAMA HARUS SAMA PERSIS DENGAN DI SQL
peta = {
    "naik_bayi": naik_bayi,
    "turun_bayi": turun_bayi,
    "naik_anak": naik_anak,
    "turun_anak": turun_anak,
    "naik_remaja": naik_remaja,
    "turun_remaja": turun_remaja,
    "naik_pemuda": naik_pemuda,
    "turun_pemuda": turun_pemuda,
    "naik_dewasa": naik_dewasa,
    "turun_dewasa": turun_dewasa,
    "naik_lansia": naik_lansia,
    "turun_lansia": turun_lansia
}

# 4. AMBIL NILAI
def hitung(usia, nama_tabel):
    cur = db.cursor()
    cur.execute(
        f"SELECT fungsi FROM {nama_tabel} WHERE %s >= b_bawah AND %s < b_atas",
        (usia, usia)
    )
    hasil = cur.fetchone()
    cur.close()

    if not hasil or hasil[0] == "0":
        return 0.0
    return round(peta[hasil[0]](usia), 4)

# 5. TAMPILKAN HASIL
daftar_usia = [1, 4, 7, 10, 12, 17, 22, 30, 50, 65, 75]

print("="*85)
print(f"{'No':<3} {'Input Usia':<10} {'Bayi':<8} {'Anak':<8} {'Remaja':<8} {'Pemuda':<8} {'Dewasa':<8} {'Lansia':<8}")
print("-"*85)

for no, u in enumerate(daftar_usia, 1):
    b = hitung(u, "bayi")
    a = hitung(u, "anak")
    r = hitung(u, "remaja")
    p = hitung(u, "pemuda")
    d = hitung(u, "dewasa")
    l = hitung(u, "lansia")
    print(f"{no:<3} {u:<10} {b:<8.4f} {a:<8.4f} {r:<8.4f} {p:<8.4f} {d:<8.4f} {l:<8.4f}")

print("="*85)
db.close()
print("\nSelesai!")
```
